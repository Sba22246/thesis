```python
import matplotlib.pyplot as plt

# Group by 'role' and count the occurrences
position_counts = df_soccer['role'].value_counts()

# Create a bar plot
plt.figure(figsize=(10, 6))
position_counts.plot(kind='bar', color='skyblue')
plt.title('Total Injuries x Different Positions')
plt.xlabel('Position')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()

# Show the plot
plt.show()
```

```python
# Group by 'role' and count the occurrences for ACL injuries
acl_position_counts = df_acl['role'].value_counts()

# Group by 'role' and count the occurrences for total injuries
position_counts = df_soccer['role'].value_counts()

# Create a figure with two subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Plot for ACL injuries
acl_position_counts.plot(kind='bar', color='skyblue', ax=ax1)
ax1.set_title('ACL Injuries x Different Positions')
ax1.set_xlabel('Position')
ax1.set_ylabel('Count')
ax1.set_xticklabels(acl_position_counts.index, rotation=45, ha='right')

# Plot for total injuries
position_counts.plot(kind='bar', color='orange', ax=ax2)
ax2.set_title('Total Injuries x Different Positions')
ax2.set_xlabel('Position')
ax2.set_ylabel('Count')
ax2.set_xticklabels(position_counts.index, rotation=45, ha='right')

# Adjust layout
plt.tight_layout()

# Show the plots
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# List of attributes to plot histograms
numeric_attributes = ['age', 'height', 'weight']

# Convert 'age' and 'height' columns to numeric
df_acl[numeric_attributes] = df_acl[numeric_attributes].apply(pd.to_numeric, errors='coerce')
df_soccer[numeric_attributes] = df_soccer[numeric_attributes].apply(pd.to_numeric, errors='coerce')

# Create a figure with subplots
fig, axes = plt.subplots(len(numeric_attributes), 2, figsize=(15, 12))

# Loop through each numeric attribute
for i, attr in enumerate(numeric_attributes):
    # Create histograms for ACL injuries and total injuries
    acl_ax = axes[i, 0]
    df_acl[attr].plot(kind='hist', bins=20, color='skyblue', ax=acl_ax)
    acl_ax.set_title(f'ACL Injuries - Histogram of {attr}')
    acl_ax.set_xlabel(attr)
    acl_ax.set_ylabel('Frequency')
    
    total_ax = axes[i, 1]
    df_soccer[attr].plot(kind='hist', bins=20, color='orange', ax=total_ax)
    total_ax.set_title(f'Total Injuries - Histogram of {attr}')
    total_ax.set_xlabel(attr)
    total_ax.set_ylabel('Frequency')

# Adjust layout
plt.tight_layout()

# Show the plots
plt.show()

```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Calculate the correlation matrix for the oversampled dataset
corr_oversampled = oversampled_df.corr()

# Generate a heatmap for the oversampled dataset
plt.figure(figsize=(12, 8))
sns.heatmap(corr_oversampled, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Matrix for Oversampled Dataset")
plt.show()

# Calculate the correlation matrix for the original dataset (assuming you have one)
corr_original = oversampled_df.corr()

# Generate a heatmap for the original dataset
plt.figure(figsize=(12, 8))
sns.heatmap(corr_original, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Matrix for Original Dataset")
plt.show()
```
