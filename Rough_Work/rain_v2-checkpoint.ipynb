```python
#https://www.arcgis.com/apps/dashboards/d29ad7056f3548eb8affeb1c0ad50106

import pandas as pd

file_path = r'C:/Users/Dell/OneDrive/Desktop/eng_rain.csv'
Eng_rain = pd.read_csv(file_path)

Eng_rain.head()

#Drop f from category
Eng_rain['category'] = Eng_rain['category'].str.replace('F', '')
#drop _ and two follwoing digits
Eng_rain['category'] = Eng_rain['category'].str[:-3]
#annual average for each year based on average monthly
Eng_rain = Eng_rain.groupby('category')['value'].mean()


Eng_rain

# Create a DataFrame from the Series
Eng_rain = Eng_rain.reset_index()

# Rename the columns
Eng_rain.rename(columns={'category': 'Year', 'value': 'avg_eng_precip'}, inplace=True)

# Convert the "Year" column to integers
Eng_rain['Year'] = Eng_rain['Year'].astype(int)

Eng_rain = Eng_rain[(Eng_rain['Year'] >= 2006) & (Eng_rain['Year'] <= 2018)]


Eng_rain

# ROW RAIN

#
file_path = r'C:/Users/Dell/OneDrive/Desktop/row_rain.csv'
row_rain = pd.read_csv(file_path)

row_rain.head()

# Transpose the pivoted DataFrame
transposed_df = row_rain.T

# Reset the index and insert a new numerical index
transposed_df.reset_index(drop=False, inplace=True)
transposed_df




# Set the first row as column headers
transposed_df.columns = transposed_df.iloc[0]

# Reset the index

transposed_df

# Rename the columns
transposed_df.rename(columns={'Country Name': 'Year',
                              'Germany': 'ger_rain',
                              'Spain': 'spa_rain',
                              'France': 'fra_rain',
                              'Italy': 'ita_rain'},
                     inplace=True)
# Drop the first row
transposed_df = transposed_df.drop(0)

transposed_df

# Convert the 'Year' column to the same data type
transposed_df['Year'] = transposed_df['Year'].astype(int)
Eng_rain['Year'] = Eng_rain['Year'].astype(int)

# Merge the DataFrames on 'Year'
annual_rain_df = pd.merge(transposed_df, Eng_rain, on='Year')

annual_rain_df


# Season Range: 2006 to 2019
```
